#### Typescript
> ⚠️ **Nota:** Para executar as células TypeScript, troque o kernel para **Deno** no canto superior direito do notebook.

````typescript
//let vs const: let permite reatribuir valores, const não permite
let inteiro: number = 24;
let decimal: number = 24.5;

console.log(inteiro, typeof inteiro);
console.log(decimal, typeof decimal);
````

````typescript
const inteiro_const: number = 24;
const decimal_const: number = 24.5;

console.log(inteiro_const, typeof inteiro_const);
console.log(decimal_const, typeof decimal_const);
````

#### React

> ⚠️ **Nota:** O React é visual, então o correspondente de variável nele seria um componente. Utilizar o Kernel **Deno** no canto superior direito do notebook.

**Exemplo 1: JSX Sintaxe Pura** (Demonstração)

```jsx

    return (
        <button>
            Hello World
        </button>
    );
}
```

Para executar React no Jupyter, veja as células abaixo com exemplos funcionais!

````typescript
// Exemplo 2: React sem JSX (funciona direto!)
import React from "npm:react@19";
import { renderToString } from "npm:react-dom@19/server";

function HelloWorld(){
    return React.createElement('button', {}, 'Hello World');
}

const html = renderToString(React.createElement(HelloWorld));
console.log("HTML gerado:", html);
````

````typescript
// Exemplo 3: Componente com Props
import React from "npm:react@19";
import { renderToString } from "npm:react-dom@19/server";

function Greeting({ name }) {
    return React.createElement('h1', {}, `Olá, ${name}!`);
}

const html = renderToString(
    React.createElement(Greeting, { name: 'Janaina' })
);


console.log("Componente renderizado:", html);
````

#### Java

````typescript
// Exemplo 3: Classe com método (sem main!)
class Pessoa {
  String nome;
  int idade;

  Pessoa(String nome, int idade) {
    this.nome = nome;
    this.idade = idade;
  }

  void apresentar() {
    System.out.println("Sou " + nome + ", tenho " + idade + " anos");
  }
}

Pessoa p = new Pessoa("Janaina", 25);
p.apresentar();
````

````typescript
// Exemplo 2: Método simples
void saudacao(String nome) {
  System.out.println("Olá, " + nome + "!");
}

saudacao("Janaina");
````

#### R
> ⚠️ **Nota:** Para executar as células R, troque o kernel para **ir** no canto superior direito do notebook.

````typescript
# Verificar tipo com typeof
inteiro <- 42L
decimal <- 3.14

print(paste("Tipo do inteiro:", typeof(inteiro)))
print(paste("Tipo do decimal:", typeof(decimal)))
````

````typescript
# Em R, numeric é o tipo padrão para números
inteiro <- 42L  # L força o tipo integer
decimal <- 3.14

cat("Inteiro:", inteiro, "- Tipo:", class(inteiro), "\n")
cat("Decimal:", decimal, "- Tipo:", class(decimal), "\n")
````

#### DotNet (C#)
> ⚠️ **Nota:** Para executar as células C#, troque o kernel para **.net-csharp** no canto superior direito do notebook.

````typescript
// Outros tipos numéricos em C#
float floatNum = 3.14f;
decimal decimalNum = 3.14m;
long longNum = 1234567890L;

Console.WriteLine($"Float: {floatNum} - Tipo: {floatNum.GetType().Name}");
Console.WriteLine($"Decimal: {decimalNum} - Tipo: {decimalNum.GetType().Name}");
Console.WriteLine($"Long: {longNum} - Tipo: {longNum.GetType().Name}");
````
